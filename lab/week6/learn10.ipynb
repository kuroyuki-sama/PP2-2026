# 10장 파일 및 예외처리

내장함수 open() 을 사용하여 파일을 입출력 시킬 수 있으며, 인자로는 open(<파일명>, <모드>, <인코딩>, ...) 등이 있다.

예외처리
- try-except 구문을 이용하여 오류 상황이 발생하였을 때, 예외 상황 코드를 발생시킬 수 있다.
- -finally 구문도 있는데, 이건 except 가 발생을 하던 안하던 무조건 실행된다.
- 예외 상황의 주된 예로는 ZeroDivisionError, AttributeError, TypeError, IndexError, EOFError 등이 있다.

```python
y = 2
print(y)
z = 0
print(z)
try:
    w = y / z
    print(w)
except ZeroDivisionError:
    print("0으로 나눔")
```

파일에 쓰기모드 (w)
- 만약 해당 파일이 존재하지 않으면 현재 경로에 파일을 추가하여 쓰기 시작함
- 해당 파일에 내용이 있다 하여도 덧씌우며 진행됨

읽고쓰기는 w+

```python
file = open("test.txt", "w")
file.write("hello world.\n")
file.close()
```

파일 닫기
- .close() 로 열려있는 파일을 닫을 수 있다.
- finally 구문을 이용하여 close() 시켜 보다 안전하게 코드를 실행시킬 수 있으며, with-as 코드를 이용하여 자동으로 파일을 닫을 수 있다.

```python
with open("test.txt", "r") as f:
    f.readline()
#코드 밖으로 빠져나갈 때 자동으로 f.close() 실행
```

파일입출력에서 자주 사용되는 문자열 메서드

1. strip(<>)
    - 해당 문자열의 좌/우에 있는 <공백>을 제거시킨다.
    - <>안에는 어떤것이든 들어갈 수 있다. (단, 문자열에 한해)
    - 오른쪽은 rstrip(), 왼쪽은 lstrip()

```python
t = "   hello world   "
print(t.strip())
```

2. split(<>)  
    - 해당 문자열 안에 <공백>을 기준으로 문자열을 나눈 후, 리스트를 반환한다.

```python
t = "h e l l o w o r l d !"
print(t.split())
```

# 디렉토리 작업 및 임의 파일 접근

디렉토리 작업

os 모듈을 이용하여 디렉토리에 접근할 수 있다.

1. os.getcwd() 
    - 현재 디렉토리의 경로를 반환한다

2. os.chdir()
    - 작업 디렉토리를 변경할 수 있다.

3. os.listdir()
    - 현재 디렉토리 안에 있는 파일들을 리스트로 반환한다.

4. os.path.isfile()  
    - 현재 경로에 있는 괄호 안의 코드가 파일인지 참거짓을 반환한다.

5. endswith() <- 얘는 os 모듈의 메서드가 아님
    - 파일의 확장자가 괄호 안의 구문으로 끝나는지 확인한다. startswith() 도 있다.

이진 파일
- 이진파일 읽기모드 : rb
- 이진파일 쓰기모드 : wb
- 이진파일 수정모드 : ab

임의 접근
- 파일에는 __파일 포인터__ 가 존재하며, 파일의 시작은 0, 끝은 EOF로 이루어져있다.
- 현재 파일에서 파일 포인터가 어디에 위치해 있는지 알려주는 메서드 : tell()
- 파일 포인터를 조작하는 함수 : seek(<int>)

```python
import os

current_dir = os.path.dirname(os.path.abspath(__file__))
filepath = os.path.join(current_dir, "words.txt")
f = open(filepath, "r+", encoding="utf-8")

string = f.read(10)
print(f"읽은 문자열 : {string}")
position = f.tell()
print(f"현재 위치 : {position}")

print(f"현재 문자열 : {string}")
position = f.seek(0)
print(f"읽은 문자열 : {string}")
f.close()
```

```python
# 이미지 복사하기

infile = open("123.png", "rb")
outfile = open("copied.png", "wb")

while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()
```

pickle 모듈
- 딕셔너리, 리스트 같은 객체들을 저장하기 위한 모듈이다.
- 이진파일 형태로 저장되며, 확장자는 .p 를 사용한다.

```python
import pickle

gameOption = {
    "Sound" : 10,
    "VideoQuality" : "Ultra",
    "Money" : 10000
}

file = open("option.p", "wb")
pickle.dump(gameOption, file)
file.close()
```

```python
import pickle

file = open("option.p", "rb")
print(pickle.load(file))
```
