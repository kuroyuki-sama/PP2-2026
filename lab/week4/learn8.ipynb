# counter 클래스 만들기
- 데이터 : 현재 숫자
- 동작 : 증가하기, 현재 값 읽어오기


```python
class Counter:
    def __init__(self, initial_count = 0):
        self.count = initial_count # 변수 생성

    def increment(self, amount = 1):
        self.count += amount
    
    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())

    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment(10)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터 : 현재 채널번호, 볼륨량, ON/OFF 여부
- 동작 : 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 클래스명 : MyTelevision


```python
class MyTelevision:
    serialNum = 0

    def __init__(self, channel = 1, volume = 0, power = False):
        self.channel, self.volume, self.power = channel, volume, power
        MyTelevision.serialNum += 1
        self.num = MyTelevision.serialNum

    def turnOn(self):
        if self.power == False:
            self.power = True
            print("전원이 켜졌습니다.\n")
        else:
            print("전원이 이미 켜져있습니다.\n")

    def turnOff(self):
        if self.power == True:
            self.power = False
            print("전원이 꺼졌습니다.\n")
        else:
            print("전원이 꺼져있습니다.\n")

    def changeChannel(self, channel):
        self.channel = channel
        print(f"채널이 {channel}번으로 변경되었습니다.")
        

    def changeVolume(self, volume):
        self.volume = volume
        print(f"볼륨이 {volume}으로 변경되었습니다.")

    def status(self):
        print(f"현재 채널 : {self.channel}\n현재 볼륨량 : {self.volume}\n전원 여부 : {self.power}")
        print(f"모델이 {self.num}개 생성되었습니다.\n")


def test_tv():
    mytv = MyTelevision(channel = 11)
    mytv.turnOn()
    mytv.status()
    mytv.turnOff()
    mytv.status()

    tv2 = MyTelevision()
    tv2.status()

if __name__ == "__main__":
    test_tv()

    
```

```python
class Remote:
    def __init__(self, tv:MyTelevision):
        self.tv = tv #self.tv 는 Mytelevision 클래스의 인스턴스가 됨

    def pressOn(self):
        self.tv.turnOn()

    def pressOff(self):
        self.tv.turnOff()

    def pressDisplayStatus(self):
        self.tv.status()

def test_remote():
    tv1 = MyTelevision()
    rc1 = Remote(tv1)
    rc1.pressOn()
    rc1.pressDisplayStatus()
    rc1.pressOff()
    rc1.pressDisplayStatus()

if __name__ == "__main__":
    test_remote()
    
```

```python
# 직접 만들어본 Person 클래스
# IndexError 발생
class Person:
    member = 4
    personlst = [] #2차원배열, 리스트로 이름, 나이 저장

    def __init__(self, name, age):
        self.__name, self.__age = name, int(age)
        self.member = Person.member
        self.personlst = Person.personlst

    def append_person(self):    
        self.personlst.append(list(self.__name, self.__age))

    def get_name(self):
        print(f"멤버는 다음과 같습니다.\n")
        for idx in range(self.member):
            print(f"{self.personlst[idx][0]}")

    def get_age(self):
        print("멤버의 나이는 다음과 같습니다.\n")
        for idx in range(self.member):
            print(f"{self.personlst[idx][1]}")


    def get_person(self):
        print(f"{"이름":<5}{"나이":<5}")
        for idx in range(self.member):
            print(f"{self.personlst[idx][0]:<5}{self.personlst[idx][1]:<5}")

def test_person():
    print("팀원의 이름과 나이를 입력하세요.")
    for _ in range(4):
        n, a = input().split()
        mb = Person(n, a)

    mb.get_name()
    mb.get_age()
    mb.get_person()

if __name__ == "__main__":
    test_person()

    
```
