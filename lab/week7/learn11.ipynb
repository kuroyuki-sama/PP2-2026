내장함수 정리하기 


1. abs() 
    - 괄호 안의 숫자의 절대값을 반환한다.
    - 정수, 부동 소수점 또한 반환 가능하며, 복소수의 경우, 복소수의 크기를 반환한다.

2. all()
    - 괄호 안의 시퀀스의 모든 항목이 참이면 True를 반환한다.
    - 여기서 참인 항목은 0이 아닌 수
    - 만약 하나라도 0이라면 False를 반환한다.

3. any()
    - 괄호 안의 시퀀스의 항목에 하나라도 참인 경우 True를 반환한다.
    - 모든 항목이 0이라면 False를 반환한다.

4. bin()
    - 이진 표현을 반환한다.
    - 0b000101 처럼 시작된다.

5. eval()   
    - 괄호 안의 수식을 분석하여 실행시킨 후 반환한다.
    eval(2**3) >> 8

6. map()
    - 반복가능한 객체의 각 항목에 주어진 함수를 적용하고, 결과값을 반환한다.
    ex)
    a, b = map(int, input().split())
    >> input받은 문자열을 공백을 기준으로 나눈 후, int형으로 변환하여 a, b에 각각 저장한다.

7. dir()    
    - 객체가 가지고 있는 변수나 함수를 출력한다.

8. complex(real, imag)
    - 복소수 객체를 반환한다. 
    - real + imag * j 값을 반환

9. filter()
    - 특정 조건을 만족하는 요소를 반환한다.
    - 첫 번째 함수가 True 값을 반환하는 요소를 filter객체로 반환한다
    ex)
    def myfilter(x): return x > 3
    result = filter(myfilter, (1,2,3,4,5,6))

10. zip()   
    - 2개의 리스트를 하나로 묶어준다. 
    ex)
    num = [1,2,3,4]
    slst = ['one', 'two', 'three', 'four']
    lst = zip(num, slst)
    > [(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four')]

이터레이터
- 반복 가능한 객체
- 예) 리스트, 딕셔너리, 문자열 등

```python
# 이터레이터 원리
class Iter:
    def __init__(self, low, high):
        self.current , self.high = low, high

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.high: raise StopIteration
        else:
            self.current += 1
            return self.current - 1

c = Iter(1, 10)
for i in c: print(i, end=" ")
```

제너레이터
- 키워드 yield 를 이용하여 __함수로부터 반복가능한 객체를 생성__ 하는 방법

```python
def Gene(low, high):
    while low <= high:
        yield low
        low += 1

for i in Gene(1, 10):
    print(i, end=" ")
```

연산자 오버라이딩
- 연산자를 메서드로 정의하는 행위
- 우리가 사용하는 문자열 + 문자열은 사실 __add__()를 통해 이루어지고있던것

```python
class Point:
    def __init__(self, x = 0, y = 0):
        self.x, self.y = x, y

p1 = Point(3, 4)
p2 = Point(5, 6)    
print(p1 + p2)
```

>> 파이선에서는 객체와 객체를 더한 결과를 출력할 수 없음

```python
class Point:
    def __init__(self, x = 0, y = 0):
        self.x, self.y = x, y

    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x, y)

    def __str__(self):
        return f"Point : ({self.x}, {self.y})"

p1 = Point(3, 4)
p2 = Point(5, 6)    
print(p1 + p2)
```

>> __add__에서 객체 + 객체의 값을 Point 클래스의 새로운 객체로 반환하여 __str__()을 작동하게 함
>> 따라서 return 에 따라 print()의 결과가 Point : x , y 가 나오게 됨.
