# counter 클래스 만들기
- 데이터 : 현재 숫자
- 동작 : 증가하기, 현재 값 읽어오기


```python
class Counter:
    def __init__(self, initial_count = 0):
        self.count = initial_count # 변수 생성

    def increment(self, amount = 1):
        self.count += amount
    
    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())

    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment(10)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터 : 현재 채널번호, 볼륨량, ON/OFF 여부
- 동작 : 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 클래스명 : MyTelevision


```python
class MyTelevision:
    def __init__(self, channel = 1, volume = 0, power = False):
        self.channel, self.volume, self.power = channel, volume, power

    def turnOn(self):
        if self.power == False:
            self.power = True
            print("전원이 켜졌습니다.\n")
        else:
            print("전원이 이미 켜져있습니다.\n")

    def turnOff(self):
        if self.power == True:
            self.power = False
            print("전원이 꺼졌습니다.\n")
        else:
            print("전원이 꺼져있습니다.\n")

    def changeChannel(self, channel):
        self.channel = channel
        print(f"채널이 {channel}번으로 변경되었습니다.")
        

    def changeVolume(self, volume):
        self.volume = volume
        print(f"볼륨이 {volume}으로 변경되었습니다.")

    def status(self):
        print(f"현재 채널 : {self.channel}\n현재 볼륨량 : {self.volume}\n전원 여부 : {self.power}")


def test_tv():
    mytv = MyTelevision(channel = 11)
    mytv.turnOn()
    mytv.status()
    mytv.turnOff()
    mytv.status()

if __name__ == "__main__":
    test_tv()

    
```

# TV 리모콘 클래스
- 데이터는 없고, 동작만 있음
- 연관된 TV를 키거나 꺼라는 동작만 지원
- TV 상태 유지 추가

```python
class Remote:
    def __init__(self, tv:MyTelevision):
        self.tv = tv #self.tv 는 Mytelevision 클래스의 인스턴스가 됨

    def pressOn(self):
        self.tv.turnOn()

    def pressOff(self):
        self.tv.turnOff()

    def pressDisplayStatus(self):
        self.tv.status()

def test_remote():
    tv1 = MyTelevision()
    rc1 = Remote(tv1)
    rc1.pressOn()
    rc1.pressDisplayStatus()
    rc1.pressOff()
    rc1.pressDisplayStatus()

if __name__ == "__main__":
    test_remote()
    
```

# 정보은닉
- 가시성 처리 방법

- private : 변수나 함수이름을 __ 로 시작한다.
 * 외부에서 접근이 불가능하다.

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def display_info(self):
        print(f"Name : {self.__name}\nAge : {self.__age}")

kim = Student("Kim", 21)
kim.display_info()
print(kim.__age)
        
```
