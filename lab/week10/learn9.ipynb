# 9장 GUI 프로그래밍

GUI 프로그래밍을 구현하는 내장모듈 : tkinter

```python
# 기본 형태
from tkinter import *
window = Tk() # tkinter 객체 생성 (루트 윈도우)
label = Label(window, text="Hello tkinter") # 화면 상단에 글씨 "생성"
label.pack() # 생성된 글씨를 "화면에 출력"

window = mainloop() # 창을 닫을 때까지 무한루프
```

* 기본 위젯들
    1. Button(부모컨테이너, text="", bg="", fg="", width="", height="", ...)
        - 클릭 가능한 버튼을 생성한다.
    
    2. Entry(부모컨테이너, fg="", bg="", width="", ...)
        - 사용자가 입력한 내용을 tkinter에 전달하여 출력한다. (한 줄만 입출력)

    3. Text(부모컨테이너, width="", ...)
        - 사용자가 입력한 내용을 tkinter에 전달하여 출력한다. (여러 줄을 입출력 가능)
        
        - insert(index, text) : 지정된 위치에 텍스트를 삽입한다.
        - get(start, end) : 지정된 위치에 텍스트를 가져온다.
        - delete(start, end) : 지정된 범위의 텍스트 삭제
    
    4. Frame()
        - 여러 위젯들을 하나로 묶어 관리하기 위한 위젯
        - 다른 위젯들을 자식 클래스로 받을 수 있다. (자신이 부모컨테이너가 됨)

```python
from tkinter import *

window = Tk()
window.geometry("1280x960")

btn = Button(window, text="버튼", width=4, height=3, bg="skyblue")
btn.pack()
entry = Entry(window, bg="yellow", width=5)
entry.pack()
text = Text(window, width=50)
text.insert("1.0", "안녕하세요!\n")
text.insert("end", "Tkinter의 Text 클래스입니다.")
text.pack()

fr = Frame(window, bg="pink")

Label(fr, text="Frame 안에 있는 라벨").pack(side="left") # 변수 없이 이렇게도 객체 생성 가능
Button(fr, text="Frame 안에 있는 버튼").pack(side="right")
Entry(fr, fg="green", width=4).pack() # 왼쪽으로 정렬하는 매개변수
fr.pack(padx=20, pady=30, fill="both", expand=True)

window.mainloop()
```

* 배치 관리자
    - 위젯들의 위치를 배치할 수 있다.

    1. pack() : 압축 배치 관리자
        - 위젯들을 사각형 블록으로 간주하여 컨테이너 안에 상하로 배치한다.
    
    2. grid() : 격자 배치 관리자
        - 위젯들을 2차원적인 격자에 배치한다.

    3. place() : 절대 배치 관리자
        - 사용자가 지정한 위치에 위젯을 배치한다.

```python
# 1. pack()
from tkinter import *

window = Tk()

Button(window, bg="red", text="박스 1").pack()
Button(window, bg="green", text="박스 2").pack()
Button(window, bg="orange", text="박스 3").pack()

window.mainloop()
```

```python
# 2. grid()
from tkinter import *

window = Tk()

btn1 = Button(window, bg="red", text="박스 1")
btn2 = Button(window, bg="green", text="박스 2")
btn3 = Button(window, bg="orange", text="박스 3")
btn4 = Button(window, bg="pink", text="박스 4")

# n행 n열로 배치
btn1.grid(row=0, column=0)
btn2.grid(row=0, column=1)
btn3.grid(row=1, column=0)
btn4.grid(row=1, column=1)

window.mainloop()
```

```python
# 3. place()
from tkinter import *

window = Tk()
window.geometry("800x600")

btn1 = Button(window, bg="red", text="박스 1")
btn2 = Button(window, bg="green", text="박스 2")
btn3 = Button(window, bg="orange", text="박스 3")

btn1.place(x=0, y=0)
btn2.place(x=60, y=80)
btn3.place(x=400, y=520)

window.mainloop()
```

버튼 이벤트 처리
    
작동 방식
1. 사용자가 버튼을 클릭한다
2. 이벤트가 발생한다
3. 코드가 실행된다.
4. 어떤 작업을 실행한다.

이때, 3에서 발생되는 함수를 **콜백함수** 라고한다.

```python
from tkinter import *

def process():
    print("버튼이 출력되었습니다.")

window = Tk()
b1 = Button(window, text="클릭하세요!", command=process)
b1.pack()

window.mainloop()
```
