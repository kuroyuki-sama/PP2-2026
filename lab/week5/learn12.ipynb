# 12장 상속

- 상속에 대한 이론과 실제 학습

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker, self.model, self.color, self.price = maker, model, color, price
    
    def setMaker(self, new_maker): self.maker = new_maker
    def getMaker(self) : return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"차량 = ({self.maker}, {self.model}, {self.color}, {self.price})"

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, batterySize):
        super().__init__(maker, model, color, price)
        self.battery = batterySize
    
    def setBattery(self, battery): self.battery = battery
    def getBattery(self): return self.battery

    def __str__(self):
        return f"{super().__str__()}, battery size = {self.battery}kWh"

class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad
    
    def __str__(self):
        return f"{super().__str__()}, MaxLoad = {self.payLoad}kg"

    def setpayLoad(self, setpay): self.payLoad = setpay
    def getpayLoad(self): return self.payLoad
    

def main():
    car1 = Car("Kia", "Ray", "Blue", 2000)
    print(car1.getDesc())
    print(car1.model)

    ecar1 = ElectricCar("Tisla", "Model S", "whlte", 10000, 81)
    ecar1.setMaker("Tesla")
    ecar1.getBattery()
    print(ecar1)
    ecar1.setBattery(92)
    print(ecar1.getDesc())

    truc1 = Truck("Ford", "F-150", "Black", 5000, 1000)
    print(truc1)
    
    print(truc1.getDesc())
    print(type(truc1))
    print(type(ecar1))
    print(type(car1))

    print(isinstance(truc1, Truck))
    print(isinstance(ecar1, ElectricCar))
    print(isinstance(car1, Car))

    print(isinstance(truc1, Car))
    print(isinstance(ecar1, Car))

    # print(isinstance(truc1, Vehicle))


if __name__ == "__main__":
    main()


```

# 속성변수의 가시성 : public, protected (_로 시작), private (__로 시작)

```python
class Parent(object):
    def __init__(self):
        self.__private_money = 1000
        self._protected_money = 500
        self.public_money = 100

    def __str__(self):
        return f"Parent has {self.__private_money + self._protected_money + self.public_money} money"


class Child(Parent):
    def __init__(self):
        super().__init__()

    def __str__(self):
        return f"Child : {self._protected_money + self.public_money} money"

def test():
    p = Parent()
    print(p)

    c = Child()
    print(c)

    print(p.public_money)
    print(p._protected_money)
    # print(p.__private_money)
    print()
    print(c.public_money)
    print(c._protected_money)
    # print(c.__private_money)
    

test()
```

# 다중 상속 : 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1 = 10):
        self.atr1 = v1

    def display1(self):
        print(f"{self.atr1}")

class Base2:
    def __init__(self, v2 = 20):
        self.atr2 = v2

    def display2(self):
        print(f"{self.atr2}")

class C(Base1, Base2):
    def __init__(self, v1, v2, v3 = 30):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.atr3 = v3

    def printstr(self): 
        print(f"{self.atr1}, {self.atr2}, {self.atr3}")
        self.display1()
        self.display2()

def test():
    obj = C(3, 5)
    obj.printstr()

test()
```

# Overriding : 부모의 함수(메소드) 재정의

```python
import math

class Shape:
    def __init__(self, name = "도형"):
        self.name = name

    def draw(self): print(f"Shape: {self.name} 을(를) 그립니다.")

    def getArea(self): print(f"Shape : {self.name}의 면적을 계산합니다.")


class Circle(Shape):
    def __init__(self, radius = 10, name = "원"):
        super().__init__(name)
        self.radius = radius

    def draw(self): 
        print(f"Circle: {self.name} 을(를) 그립니다.")

    def getArea(self): 
        print(f"Cirlce: {self.name}의 면적을 계산합니다.")
        return round(math.pi * self.radius ** 2, 2)

class Rectangle(Shape):
    def __init__(self, width = 1, height = 1, name="사각형"):
        super().__init__(name)
        self.width, self.height = width, height

    def draw(self): print(f"Rectangle: {self.name} 을(를) 그립니다.")

    def getArea(self): 
        print(f"Rectangle : {self.name}의 면적을 계산합니다.")
        return self.height * self.width

def test():
    c = Circle(1)
    c.draw()
    print(c.getArea())

    r1 = Rectangle(5, 10)
    r1.draw()
    print(r1.getArea())

    # print("-" * 20)
    # circles = [Circle(1), Circle(2), Circle(3)]
    # for circle in circles:
    #     # circle.draw()
    #     print(f"반지름 : {circle.radius}")
    #     print(circle.getArea())
    #     print()

    # print("=" * 20)
    # rect = [Rectangle(1, 2), Rectangle(3, 4), Rectangle(5, 6)]
    # for rec in rect:
    #     print(f"가로 : {rec.width}, 세로 : {rec.height}")
    #     print(rec.getArea())
    #     print()

    print("-=" * 10)
    shapes = [Circle(1), Rectangle(1,2), Circle(20), Rectangle(2, 3)]
    for s in shapes:
        print(s.getArea())
        print()
if __name__ == "__main__":
    test()
```
