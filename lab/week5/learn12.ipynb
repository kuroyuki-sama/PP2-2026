# 12장 상속

- 상속에 대한 이론과 실제 학습

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker, self.model, self.color, self.price = maker, model, color, price
    
    def setMaker(self, new_maker): self.maker = new_maker
    def getMaker(self) : return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"차량 = ({self.maker}, {self.model}, {self.color}, {self.price})"

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, batterySize):
        super().__init__(maker, model, color, price)
        self.battery = batterySize
    
    def setBattery(self, battery): self.battery = battery
    def getBattery(self): return self.battery

    def __str__(self):
        return f"{super().__str__()}, battery size = {self.battery}kWh"

class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad
    
    def __str__(self):
        return f"{super().__str__()}, MaxLoad = {self.payLoad}kg"

    def setpayLoad(self, setpay): self.payLoad = setpay
    def getpayLoad(self): return self.payLoad
    

def main():
    car1 = Car("Kia", "Ray", "Blue", 2000)
    print(car1.getDesc())
    print(car1.model)

    ecar1 = ElectricCar("Tisla", "Model S", "whlte", 10000, 81)
    ecar1.setMaker("Tesla")
    ecar1.getBattery()
    print(ecar1)
    ecar1.setBattery(92)
    print(ecar1.getDesc())

    truc1 = Truck("Ford", "F-150", "Black", 5000, 1000)
    print(truc1)
    
    print(truc1.getDesc())
    print(type(truc1))
    print(type(ecar1))
    print(type(car1))

    print(isinstance(truc1, Truck))
    print(isinstance(ecar1, ElectricCar))
    print(isinstance(car1, Car))

    print(isinstance(truc1, Car))
    print(isinstance(ecar1, Car))

    # print(isinstance(truc1, Vehicle))


if __name__ == "__main__":
    main()


```
